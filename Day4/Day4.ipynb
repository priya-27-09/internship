```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Priya", "Selva", "Surekha", "Meena"],
    "Department": ["AIML", "CSE", "AIML", "CSE"],
    "Marks": [90, 85, np.nan, 75],
    "Age": [19, 20, 19, np.nan]
}

df = pd.DataFrame(data)

# Check Missing Values
print("Missing Values:")
print(df.isnull().sum())

# Fill Missing Values
df["Marks"] = df["Marks"].fillna(df["Marks"].median())
df["Age"] = df["Age"].fillna(df["Age"].median())

print("\nAfter Filling Missing Values:")
print(df)

# Drop Missing Values
df_clean = df.dropna()

print("\nCleaned Data:")
print(df_clean)
```

```python
import pandas as pd
import numpy as np
data = {
    "Name": ["Priya", "Selva", "Surekha", "Malavika",
             "Meena", "Medhi", "Arun", "Kavi", "Ravi", "Divya"],
    "Department": ["AIML", "CSE", "AIML", "CSE", "AIML",
                   "CSE", "AIML", "CSE", "AIML", "CSE"],
    "Age": [19, 20, 19, 20, 19, np.nan, 21, 20, 21, 19],
    "Marks": [90, 85, 95, 88, np.nan, 75, 80, 92, 70, 85]
}
df = pd.DataFrame(data)
df.to_csv("students.csv", index=False)
df = pd.read_csv("students.csv")
print("ORIGINAL DATA:")
print(df)
print("\nMissing Values:")
print(df.isnull().sum())
df["Marks"] = df["Marks"].fillna(df["Marks"].median())
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Grade"] = df["Marks"].apply(
    lambda x: "A" if x >= 90 else
              "B" if x >= 80 else
              "C" if x >= 70 else "D"
)
print("\nCLEANED DATA:")
print(df)
print("\nTotal Students:", len(df))
print("Average Marks:", df["Marks"].mean())
print("\nTopper:")
print(df.loc[df["Marks"].idxmax()])
print("\nDepartment-wise Average:")
print(df.groupby("Department")["Marks"].mean())
print("\nStudents scoring 85 or above:")
print(df[df["Marks"] >= 85])
print("\nSorted Marks:")
print(df.sort_values("Marks", ascending=False))
df.to_csv("clean.csv", index=False)
print("\nCleaned Dataset Saved Successfully!")
```
